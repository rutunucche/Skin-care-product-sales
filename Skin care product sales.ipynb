```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
sns.countplot(data=sales, x="Sales_Person", hue="Product", palette="Set2")

plt.title("Person vs Product", size=25)
plt.xlabel("Person", size=20)
plt.ylabel("Count", size=20)
plt.xticks(size=15,rotation=75)
plt.legend(title="Product", fontsize=12)
plt.show()
```

```python
plt.hist(
    x="Amount",  #column
    data=sales,    #dataset
    bins=300,   #divide range in 10 equal width
    color="red"  #color
)

plt.title("amount")    #title
plt.xticks(np.arange(0,20), rotation=75)  #axis from 0 to 9 with rotation 75
plt.show()
```

```python
plt.figure(figsize=(20,10))

#first
plt.subplot(141) # 1 row, 4 column, 1 position
plt.hist(x="Country",data=sales, bins=10, color="orange")    
plt.title("Country")                                 
plt.xticks(rotation=90)

#second
plt.subplot(142)         
plt.hist(x="Product", data=sales, bins=10, color="red")
plt.title("Product")     
plt.xticks(rotation=90)

plt.show()
```
